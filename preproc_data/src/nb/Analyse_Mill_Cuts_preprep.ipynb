```python
tool_type = 'C1030'

from pathlib import Path
import pandas as pd

base = Path(f"../../../data/raw/{tool_type}")
print(f"Base folder: {base.absolute()}")
# Target output folder
out_folder = base / "all"
out_folder.mkdir(parents=True, exist_ok=True)

# Column rename dictionary (example!)
	

rename_map = {
    '+/Nck/!SD/nckServoDataActCurr32 [u1; 1]': 'iqAx1',
    '+/Nck/!SD/nckServoDataActCurr32 [u1; 2]': 'iqAx2',
    '+/Nck/!SD/nckServoDataActCurr32 [u1; 3]': 'iqAx3',
    '+/Nck/!SD/nckServoDataActCurr32 [u1; 4]': 'iqAx4',
    '+/Nck/!SD/nckServoDataActPos1stEnc32 [u1; 1]': 'posAx1',
    '+/Nck/!SD/nckServoDataActPos1stEnc32 [u1; 2]': 'posAx2',
    '+/Nck/!SD/nckServoDataActPos1stEnc32 [u1; 3]': 'posAx3',
    '+/Channel/!RP/rpa [u1; 11]': 'rpa',
    '+/Nck/!SD/nckServoDataActPower32 [u1; 4]': 'powerAx4',
    '+/Nck/!SD/nckServoDataActVelMot32 [u1; 4]': 'velAx4'
    # add more …
}

# Iterate through all Vxx folders
for folder in sorted(base.glob("V*")):
   
    if folder.is_dir():
        print(f"Processing folder: {folder.name}")

        for csv_file in folder.glob("*.csv"):
            df = pd.read_csv(csv_file)

            # Validate & rename
            cols_to_rename = {c: rename_map[c] for c in df.columns if c in rename_map}
            df = df.rename(columns=cols_to_rename)
            # Avoid overwrite: prefix with Vxx
            new_name = f"{csv_file.name}"
            df.to_csv(out_folder / new_name, index=False)

print("Done. All renamed files stored in /all")
```

```python
df = exp_params_df.copy()
results = []
tool_type = 'C1030'  # insert tool type here C4240
#param_experim_all_files = f'../../../data/parameterize_experiment/Experiment_{tool_type}.xlsx'
file_folder = f'../../../data/raw/{tool_type}/all/' #f'../../../data/raw/{tool_type}/all/'
#C:\Users\thmru\github\2025_LM_FinalProject\data\raw\C1030\all
N_csv_files = len(df)
print(f"Total number of measurement files to process: {N_csv_files}")
# take a random sample of files to plot
idx_random = np.random.choice(N_csv_files, size=3, replace=False)
# Loop over each measurement file
for idx, row in df.iterrows():

    # 1. Read real measurement file
    file_path = file_folder + row["SinuTraceFile (*.csv)"]
    #print(f"Processing measurement {idx}: {file_path}")
    real = pd.read_csv(file_path)
    t = real["time"].values
    #print(f"Time data length: {len(t)}, data tpye: {type(t)}")
    iq_real = real["iqAx4"].values

    # 2. Extract parameters
    ap = row["Ap (mm)"]
    fz = row["fz (mm/tooth)"]                     # already in dataframe
    z  = int(row["z"])
    kappa = row["kappa (rad)"]
    omega = row["omega (rad/s)"]
    kc11 = row["kc11"]
    mc = row["mc"]
    rtool = row["rtool (m)"]
    phi_ent = row["phi_ent_rad"]
    phi_exit = row["phi_exit_rad"]

    #3. find all values with same sample rate
    mill_preprocess = PreprocessData(fs=500)
    mill_preprocess.set_time(t)
    idx_s, idx_e = mill_preprocess.find_constant_fs_region()
    if len(t) - idx_e > 100:
        print(f"Using data from index {idx_s} to {idx_e}, total points: {idx_e - idx_s}, length of t: {len(t)} ")
    # 3. Calculate model torque on same time stamps
    mill_model = KienzleMillingModel( ap, fz, z, kappa, omega, kc11, mc, rtool, phi_ent, phi_exit)
    phi, Fc, Tc, sin_sums = mill_model.evaluate(t[idx_s:idx_e])

    # preprocessing: only in cut parameters
    mill_preprocess.set_real_signal(iq_real[idx_s:idx_e])
    mill_preprocess.set_model_signal(Tc)
    tau, lag_samples, cut_idx = mill_preprocess.estimate_delay()
    print(f"Measurement {idx}: Estimated delay (tau): {tau:.6f} s, lag samples: {lag_samples}")
    Tc_model_shifted = mill_preprocess.shift_signal(Tc, lag_samples)
    # 4. Create synchronized output dataframe
    out = pd.DataFrame({
        "time": t[cut_idx:idx_e],
        "Tc_model": Tc[cut_idx:idx_e],
        "Tc_model_shifted": Tc_model_shifted[cut_idx:idx_e],
        "Iq_real": iq_real[cut_idx:idx_e],
        "phi": phi[cut_idx:idx_e],
        "sin_sums": sin_sums[cut_idx:idx_e],
        "measurement_id": idx
    })

    # 5. Save result file per measurement
    fname =row["SinuTraceFile (*.csv)"]
    fname = f"../../../data/preprocessed/Mill/{fname}"
    out.to_csv(fname, index=False)

    results.append(fname)
    ## plot for the report
    if idx in idx_random:
        
        # measurements to plot
        plt.figure(figsize=(10,6))
        plt.plot(t, iq_real, label="Measured Iq", alpha=0.7)
        plt.xlabel("Time (s)")
        plt.ylabel("Torque / Current")
        plt.title(f"Measurement {idx} - Tool {tool_type}")
        plt.legend()
        plt.grid()
        plt.show()

        # plot with fixed samlpe rate        
        fig, axs = plt.subplots(1,2, figsize=(10, 8), sharex=True)

        
        axs[0].plot(out["time"], mill_preprocess.normalize(out["Iq_real"]), label="Measured Iq", alpha=0.7)
        axs[0].plot(out["time"], mill_preprocess.normalize(out["Tc_model"]), label="Modeled Tc (unshifted)", alpha=0.7)
        axs[0].set_ylabel("Torque / Current")
        axs[0].set_title(f"Measurement {idx} - Tool {tool_type} (unshifted model)")
        axs[0].legend()
        axs[0].grid()

        axs[1].plot(out["time"], mill_preprocess.normalize(out["Iq_real"]), label="Measured Iq", alpha=0.7)
        axs[1].plot(out["time"], mill_preprocess.normalize(out["Tc_model_shifted"]), label="Modeled Tc (shifted)", alpha=0.7)
        axs[1].set_xlabel("Time (s)")
        axs[1].set_ylabel("Torque / Current")
        axs[1].set_title(f"Measurement {idx} - Tool {tool_type} (shifted model)")
        axs[1].legend()
        axs[1].grid()

        plt.tight_layout()
        plt.show()

results[:5]   # show first few generated files
```

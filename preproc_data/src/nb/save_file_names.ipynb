jsut read the file names and save them in a Excel file

```python
import pandas as pd
from pathlib import Path

tool_type = 'C4240' #C1030'
folder = Path(f"../../../data/raw/{tool_type}/Plan")

try:
    # Explicit existence check to avoid masking errors
    if not folder.exists():
        raise FileNotFoundError(f"Folder does not exist: {folder}")

    # Read all .csv files sorted by modification time
    files = sorted(folder.glob("*.csv"), key=lambda p: p.stat().st_mtime)

    if not files:
        print(f"No .csv files found in: {folder}")
    else:
        df = pd.DataFrame({
            "filename": [f.name for f in files],
            "full_path": [str(f.resolve()) for f in files],
            "modified": [pd.to_datetime(f.stat().st_mtime, unit='s') for f in files],
        })

        df.to_excel(f"csv_file_list_{tool_type}.xlsx", index=False)
        print(f"Saved csv_file_list.xlsx with {len(df)} entries.")

except FileNotFoundError as e:
    print(e)

except Exception as e:
    # Only catch unexpected errors here; do NOT hide all logic errors
    print("Unexpected error:", e)
```
